```python
# 1)Electricity Bill Calculator

customer_name = input()
units_consumed = int(input())
if units_consumed < 0:
    print("\nInvalid units")
else:
  electricity_charge=0
  units=units_consumed
  if units > 100:
    electricity_charge +=100*2
    units -=100
  else:
    electricity_charge +=units*2
    units=0
  if units > 100:
    electricity_charge +=100*3
    units -=100
  else:
    electricity_charge +=units*3
    units=0
  if units > 200:
    electricity_charge +=200*5
    units -=200
  else:
    electricity_charge +=units*5
    units=0
  if units > 0:
    electricity_charge += units * 7
  service_charge = 50
  total_bill = electricity_charge + service_charge
  print(f"\nCustomer: {customer_name}")
  print(f"Units Consumed: {units_consumed}")
  print(f"Electricity Charge: {electricity_charge}")
  print(f"Service Charge: {service_charge}")
  print(f"Total Bill: {total_bill}")
```

```python
# 2)Student Result Management

n = int(input("Number of students: "))
passed_count=0
failed_count=0
highest_average=-1.0
top_performers=""
for i in range(n):
    print(f"\nEnter details for student {i + 1}:")
    try:
      line=input().split()
      if len(line) < 4:
            print()
            failed_count += 1
            continue
      name = line[0]
      python_marks = float(line[1])
      database_marks = float(line[2])
      aptitude_marks = float(line[3])
      if python_marks < 0 or database_marks < 0 or aptitude_marks < 0:
        print(f"Invalid marks for {name}")
        failed_count += 1
        continue
      average = (python_marks + database_marks + aptitude_marks) / 3
      result = ""
      grade = ""
      if python_marks >= 40 and database_marks >= 40 and aptitude_marks >= 40 and average >= 50:
        result = "Passed"
        passed_count += 1
        if average >= 80:
          grade = "A"
        elif average >= 70:
          grade = "B"
        elif average >= 60:
          grade = "C"
        else:
          grade = "D"
        print(f"{name} - Average: {average:.2f} - {result} - Grade: {grade}")
      else:
        result = "Failed"
        failed_count += 1
        print(f"{name} - Average: {average:.2f} - {result}")
      if average > highest_average:
        highest_average = average
        top_performers = name
      elif average == highest_average:
        if top_performers:
          top_performers += ", " + name
      else:
        top_performers = name
    except ValueError:
        print("Invalid input for marks ")
        failed_count += 1
        continue
    except IndexError:
        print("Invalid input format ")
        failed_count += 1
        continue
print(f"\nPassed: {passed_count}")
print(f"Failed: {failed_count}")
print(f"Top Performer: {top_performers}")
```

```python
# 3)Online Shopping Cart

class Product:
    def __init__(self, product_id, name, price, stock):
        self.product_id = product_id
        self.name=name
        self.price=int(price)
        self.stock=int(stock)
class ShoppingCart:
    def __init__(self):
        self.cart_items={}
    def add_product(self, product_id, quantity, catalog):
        if product_id not in catalog:
            print("Product not found in catalog")
            return
        if quantity <=0:
            return
        product=catalog[product_id]
        already_in_cart=self.cart_items.get(product_id, 0)
        if already_in_cart+quantity>product.stock:
            print("Insufficient stock")
        else:
            self.cart_items[product_id]=already_in_cart+quantity
```
